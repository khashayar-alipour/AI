# 🧠 Neural Networks (Artificial Neural Networks)

Artificial Neural Networks (**ANNs**) are one of the most important concepts in **Artificial Intelligence (AI)** and **Machine Learning (ML)**.

Modern AI systems such as:

- 🤖 ChatGPT
- 🖼️ Image Generation Models
- 👁️ Computer Vision
- 🎙️ Speech Recognition
- 🌍 Machine Translation
- 🚗 Self-driving Cars

are all built upon different types of neural networks.

---

# 💡 The Main Idea Behind Neural Networks

Traditional programming works like this:

```text
Input
   ↓
Program (Rules Written by Humans)
   ↓
Output
```

The programmer must explicitly define every rule.

For example, to detect a cat, we might write rules such as:

- Has two ears
- Has whiskers
- Has a tail
- Has four legs

Unfortunately, real-world problems are far too complex for handwritten rules.

Instead, Neural Networks learn these rules automatically from data.

The idea is simple:

> **Instead of programming the rules, let the computer learn them from examples.**

---

# 🧠 Inspiration from the Human Brain

Neural Networks were inspired by biological neurons inside the human brain.

A biological neuron:

```text
Receive Signals
      ↓
Process Information
      ↓
If Activated
      ↓
Send Signal to Other Neurons
```

Artificial neurons imitate this behavior mathematically.

Although inspired by neuroscience, modern neural networks are mathematical models rather than realistic simulations of the brain.

# 📥 Input Layer

The **Input Layer** is the first layer.

It:

- Stores the input data
- Contains one neuron for each input feature
- Performs no computation

Example:

```text
Area = 120

Rooms = 3

Age = 5

Floor = 4
```

---

# 🔍 Hidden Layers

The middle layers are called **Hidden Layers**.

These are where learning actually happens.

Each hidden layer:

- Receives inputs
- Applies weights
- Adds biases
- Uses an activation function
- Passes the output to the next layer

The more hidden layers a network has, the **deeper** it becomes.

---

# 📤 Output Layer

The final layer produces the prediction.

### Regression

Usually contains **one output neuron**.

Example:

`Predicted House Price = 4.8 Billion`

### Classification

Usually contains **one output neuron per class**.

---

# 🔗 Connectivity

## Fully Connected Network

Every neuron connects to every neuron in the next layer.

```text
○ ○ ○

↓↓↓↓↓↓↓↓

○ ○ ○ ○
```

---

## Partially Connected Network

Only selected neurons are connected.

This reduces computation and is common in architectures such as CNNs.

---

# ⚙️ How Does an Artificial Neuron Work?

Each artificial neuron performs three simple steps:

1. Receive inputs
2. Perform a mathematical computation
3. Produce an output

---

## Inputs

The neuron receives input values:

`x₁  x₂  ...  xₙ`

---

## Weights

Each input has its own weight:

`w₁  w₂  ...  wₙ`

Weights determine how important each feature is.

---

## Weighted Inputs

Each input is multiplied by its weight.

`w₁x₁  w₂x₂  ...  wₙxₙ`

---

# 🎯 Weight

A **Weight** determines how much influence an input has on the final prediction.

For example,

`Price = Area×W₁ + Rooms×W₂ + Age×W₃`

If:

`W₁ = 20`

then **Area** has a strong influence.

If:

`W₃ = -2`

then increasing the building's age decreases the predicted price.

Initially, weights are assigned random values.

During training, they are gradually adjusted to reduce prediction errors.

---

# ➕ Bias

Besides weighted inputs, every neuron adds another parameter called **Bias**.

The **linear** computation is:

`z = WᵀX + B`

or equivalently,

`z = w₁x₁ + w₂x₂ + ... + wₙxₙ + b`

Bias acts like the intercept in linear regression.

Without Bias, the neuron's decision boundary would always pass through the origin, making the model much less flexible.

---

# 🧮 Mathematical Formula of a Neuron

Every neuron computes

`z = w₁x₁ + w₂x₂ + ... + wₙxₙ + b`

where

- **x** = Inputs
- **w** = Weights
- **b** = Bias
- **z** = Weighted Sum

This value is then passed to an **Activation Function**.

# 🚀 Forward Path in Neural Networks

## 🧠 What is Forward Path?

**Forward Path** (also called **Forward Propagation** or **Forward Pass**) is the process where input data moves through a neural network from the **input layer** to the **output layer** to produce a prediction.

In simple words:

> The neural network receives input data → processes it layer by layer → produces an output.

During the forward path, the network only **makes a prediction**.  
It does **not learn or update its weights**.

---

# 🔄 The Basic Idea

A neural network has several layers:

```
Input Layer
      ↓
Hidden Layer 1
      ↓
Hidden Layer 2
      ↓
Output Layer
```

The information always moves in one direction:

`Input → Hidden Layers → Output`

That is why it is called:  **Forward Path** ➡️

---

# 🏗️ Step-by-Step Forward Path

Let's imagine a simple neural network:

```
       Input              Hidden              Output

        x1  ─────┐
                 │
        x2  ─────┼──→  Neurons  ───→  Prediction
                 │
        x3  ─────┘
```

---

# 1️⃣ Receiving Input Data

The neural network receives input features.

Example:

Predicting house prices:

```
x1 = House size
x2 = Number of rooms
x3 = Location
```

These values enter the **Input Layer**.

Important:

The input layer does not perform calculations.

It only passes data to the next layer.

---

# 2️⃣ Applying Weights

Each connection between neurons has a weight.

Example:  `Input: x1=5  |  Weight: w1=0.8`

The neuron calculates:  `x1 × w1`

For multiple inputs:  `[x_1w_1 + x_2w_2 + x_3w_3]`

The network decides how important each input is using weights.

---

# 3️⃣ Adding Bias

After multiplying inputs by weights, the neuron adds a value called **Bias**.

Formula:  `[z = W^TX + b]`

Where:

- 🟦 **X** → Input values
- 🟩 **W** → Weights
- 🟨 **b** → Bias
- 🟥 **z** → Weighted sum

Example:  `z = (x1*w1) + (x2*w2) + b`

---

# 4️⃣ Applying Activation Function

The weighted sum is passed through an activation function.

Why?

Because without activation functions, the neural network would only perform linear calculations.

# 5️⃣ Passing Information to the Next Layer

The output of one neuron becomes the input of the next neuron.

Example:

```
Layer 1

Neuron output:
0.75

        ↓

Layer 2

Neuron receives:
0.75
```

This process repeats through all hidden layers.

---

# 6️⃣ Producing Final Output

After passing through all layers, the final layer produces the prediction.

Examples:

---

## Regression

Predicting house price:  `Output: 350000 $`

One neuron is usually enough.

---

## Binary Classification

Example:

Is this email spam?

Output:  `0.92`

Meaning: `92% probability of spam`

---

## Multi-Class Classification

Example:

Image classification:

```
Cat       → 0.1
Dog       → 0.8
Car       → 0.1
```

The highest value is selected:  `Prediction = Dog`

---

# 🧮 Mathematical View of Forward Path

For one neuron:  `[z = W^TX+b]`

Then:  `[a=f(z)]`

Where:

- **z** → Weighted sum
- **f** → Activation function
- **a** → Neuron output (activation)

---

For a complete neural network:

### Layer 1:  

`[a_1=f(W_1X+b_1)]`


### Layer 2:  

`[a_2=f(W_2a_1+b_2)]`


### Output Layer:  

`[y=f(W_3a_2+b_3)]`

The final value is the network prediction.

# 🔁 Forward Path vs Backpropagation

| Forward Path ➡️ | Backpropagation ⬅️ |
|---|---|
| Makes predictions | Learns from mistakes |
| Input moves forward | Error moves backward |
| Uses current weights | Updates weights |
| Happens during training and prediction | Happens only during training |
| Calculates output | Minimizes loss |

# ✅ Important Points

- Forward Path is the **prediction process** of a neural network.
- Data always moves from **input → output**.
- Weights and biases are used to transform data.
- Activation functions add non-linearity.
- No learning happens during forward propagation.
- Learning happens later using **Backpropagation** and **Gradient Descent**.


# ⚡ Activation Function

After computing **z**, the neuron does **not** directly output it.

Instead,

```text
Inputs
    ↓
Weighted Sum
    ↓
Activation Function
    ↓
Neuron Output
```

---

# ❓ Why Do We Need Activation Functions?

Without activation functions, every neuron performs only a linear operation:

`y = wx + b`

Even if we stacked hundreds of layers together, the entire network would still behave like a single linear transformation.

That means it would be unable to learn complex nonlinear relationships.

Activation Functions introduce **nonlinearity**, allowing neural networks to solve problems such as:

- Face Recognition
- Speech Recognition
- Language Translation
- Medical Diagnosis
- Autonomous Driving

---



# 🎯 Popular Activation Functions

## 1️⃣ Sigmoid

```text
          1
σ(x) = ----------
       1 + e⁻ˣ
```

Output range:

```text
0 → 1
```

Commonly used for:

- Binary Classification
- Probability estimation

Limitation:

- Suffers from the **Vanishing Gradient Problem**.

---

## 2️⃣ Tanh

Output range:

```text
-1 → 1
```

Advantages:

- Zero-centered output
- Often performs better than Sigmoid in hidden layers

---

## 3️⃣ ReLU (Rectified Linear Unit)

```text
ReLU(x) = max(0, x)
```

Example:

```text
-7 → 0

5 → 5
```

Advantages:

- Very fast
- Computationally efficient
- Reduces the Vanishing Gradient problem
- Most widely used activation function in modern deep learning

---

## 4️⃣ Leaky ReLU

Instead of outputting zero for negative values, Leaky ReLU keeps a small slope.

```text
x > 0  → x

x ≤ 0 → 0.01x
```

This helps prevent the **Dying ReLU Problem**.

---

## 5️⃣ Softmax

Softmax converts raw scores into probabilities.

Example:

Before Softmax:

```text
Cat   2.3

Dog   5.4

Bird  1.1
```

After Softmax:

```text
Cat   0.04

Dog   0.92

Bird  0.04
```

Properties:

- Outputs are probabilities.
- All probabilities sum to **1**.
- Commonly used in multi-class classification.

# 🚀 How Does a Neural Network Learn?

Training follows these steps:

1. 📥 Input data enters the network.
2. ⚙️ Every neuron computes **Wx + b**.
3. ⚡ Activation functions generate neuron outputs (**Forward Propagation**).
4. 📊 The prediction is compared with the true label using a **Loss Function**.
5. 🔁 **Backpropagation** computes gradients.
6. 📉 **Gradient Descent** updates the weights and biases.
7. 🔄 This process repeats over many epochs until the prediction error becomes as small as possible.   


---


# 🔄 Backpropagation in Neural Networks

## 🧠 What is Backpropagation?

**Backpropagation** (short for **Backward Propagation of Errors**) is the main algorithm that allows a neural network to **learn from its mistakes**.

A neural network can perform predictions using **Forward Propagation**, but it needs a method to understand:

- ❌ How wrong was the prediction?
- 🔍 Which weights caused this error?
- 🔧 How should weights and biases be adjusted?

Backpropagation solves this problem by calculating how much each parameter contributed to the final error and then updating them to improve future predictions.

---

# 💡 Main Idea Behind Backpropagation

The learning process of a neural network is similar to learning from mistakes:

```text
Make a prediction

        ↓

Compare prediction with the correct answer

        ↓

Measure the error

        ↓

Find the source of the error

        ↓

Adjust weights and biases

        ↓

Make a better prediction next time
```

The goal is:

> 🎯 Minimize the error between the network's prediction and the actual result.

---

# 🔄 Forward Propagation vs Backpropagation

| Forward Propagation ➡️ | Backpropagation ⬅️ |
|---|---|
| Moves from input layer to output layer | Moves from output layer back through the network |
| Produces the prediction | Calculates how to improve the prediction |
| Uses current weights and biases | Updates weights and biases |
| Happens before calculating the error | Happens after calculating the error |
| Answers: "What is the output?" | Answers: "How should the network learn?" |

---


# ⬅️ Backpropagation Process

After forward propagation, the network compares the prediction with the real answer.

Example:

`Network Prediction: Dog = 0.80`

`Actual Answer: Cat`

The network calculates how incorrect its prediction is using a **Loss Function**.

The loss value represents the quality of the prediction.

Example:

```text
Large Loss

↓

Prediction is far from reality


Small Loss

↓

Prediction is closer to reality
```

---

# 🧩 Steps of Backpropagation

## 1️⃣ Calculate the Error

The first step is measuring the difference between:

- Predicted output
- Actual output

Example:

`Prediction: 0.80`

`Actual: 1.00`

`Error: 0.20`

The purpose of training is to reduce this error.

---

# 2️⃣ Calculate the Gradient

The network needs to know:

> Which weights had the biggest effect on the error?

For this, backpropagation uses derivatives from calculus.

It calculates:  `[ frac{\partial Loss}{\partial Weight} ]`

This value is called the **Gradient**.

The gradient tells us:

- The direction of the required change
- How much each weight affects the loss

---

# 3️⃣ Update Weights and Biases

After calculating gradients, the network updates its parameters.

The update rule is:  `New\ Weight = Old\ Weight - Learning\ Rate \times Gradient`

Where:

## Learning Rate

The learning rate controls the size of each update.

### Small Learning Rate:

```text
Small updates

✅ More stable learning

❌ Slower training
```

### Large Learning Rate:

```text
Large updates

✅ Faster training

❌ Risk of missing the optimal solution
```

---

# 🔁 How Does Error Move Backward?

During forward propagation:

```text
Input

 ↓

Hidden Layers

 ↓

Output
```

During backpropagation, the error moves in the opposite direction:

```text
Output Layer

      ↓

Hidden Layer

      ↓

Hidden Layer

      ↓

Input Layer
```

The algorithm starts from the output because the final error is known there.

Then it calculates how much each previous layer contributed to that error.

---

# 🔗 Backpropagation and Gradient Descent

These two concepts are related but different.

## Backpropagation

Backpropagation calculates:

> "How much should each weight change?"

It produces gradients.

---

## Gradient Descent

Gradient Descent performs the actual update:

> "Change the weights in the direction that reduces the error."

Together:

```text
Forward Propagation

        ↓

Prediction

        ↓

Loss Function

        ↓

Backpropagation

        ↓

Calculate Gradients

        ↓

Gradient Descent

        ↓

Update Weights

        ↓

Repeat
```

---

# 🚀 Training Procedure in Neural Networks

Training a Neural Network is the process of teaching the model how to perform a specific task by adjusting its internal parameters (**weights and biases**) based on data.

The main goal of training is:

> 🎯 Find the best parameters that minimize the prediction error and allow the network to generalize well on unseen data.

A neural network does not learn by receiving rules. Instead, it learns patterns by repeatedly making predictions, measuring its mistakes, and updating its parameters.

---

# 🧠 Complete Neural Network Training Procedure

The entire training process can be summarized as:

```text
Training Dataset

      ↓

Data Processing

      ↓

Parameter Initialization

      ↓

Training Loop

      ↓

Forward Propagation

      ↓

Loss Calculation

      ↓

Backpropagation

      ↓

Parameter Update

      ↓

Repeat Until Training is Complete
```

---

# 1️⃣ Prepare the Training Data

Before training begins, the dataset must be prepared.

A dataset usually contains:

- **Input Features (X)** → Information given to the network
- **Target Values / Labels (y)** → Correct answers that the network should learn


The dataset is usually divided into three parts:

## Training Set

Used to train the model.

```text
The model learns patterns from this data.
```

## Validation Set

Used during training to monitor the model's performance.

```text
Helps detect problems such as overfitting.
```

## Test Set

Used after training is finished.

```text
Measures the final performance on unseen data.
```

---

# 2️⃣ Initialize Network Parameters

Before learning starts, the network needs initial values for its parameters.

These parameters include:

- Weights
- Biases

Usually:

- Weights are initialized with small random values.
- Biases are initialized with small values.

Example:

```text
Weight 1 = 0.25

Weight 2 = -0.14

Bias = 0.01
```

At this stage, the network does not understand the problem.

These values are only initial guesses that will be improved during training.

---

# 3️⃣ Training Loop

The **Training Loop** is the repeated process that allows the neural network to learn.

During each loop, the network performs these steps:

```text
1. Take a batch of training data

2. Generate predictions

3. Calculate the error

4. Calculate gradients

5. Update weights and biases

6. Repeat
```

The training loop continues many times until the model reaches acceptable performance.

Conceptually:

```python
for batch in dataset:

    prediction = model(batch)

    loss = calculate_loss(prediction)

    gradients = backpropagation(loss)

    update_parameters(gradients)
```

---

# 4️⃣ Batch Processing

A neural network usually does not process the entire dataset at once.

Instead, the data is divided into smaller groups called **batches**.

Example:

Dataset:  `100,000 images`

Batch size:  `100 images`

The training process becomes:

```text
Batch 1

↓

Prediction

↓

Calculate Error

↓

Update Weights


Batch 2

↓

Prediction

↓

Calculate Error

↓

Update Weights


...

Batch 1000
```

Each batch causes one parameter update.


---

# 5️⃣ Calculate Loss

After producing a prediction, the network compares it with the real answer.

The difference between prediction and reality is measured using a **Loss Function**.

Example:

Prediction: `$450,000`

Actual Value: `$500,000`

Loss: `Large Error`

The loss value tells the network how far its prediction is from the desired output.

The goal of training is to minimize this loss.

---

# 6️⃣ Backpropagation

After calculating the loss, the network needs to understand:

- Which parameters caused the error?
- How much should each parameter change?

This is done using **Backpropagation**.

Backpropagation sends the error information backward through the network:

```text
Output Layer

      ↑

Hidden Layer

      ↑

Hidden Layer

      ↑

Input Layer
```

It calculates the **gradient** of the loss with respect to each parameter.

The gradient shows:

```text
How much does this parameter affect the error?
```

Mathematically: `frac{\partial Loss}{\partial Weight}`

---

# 7️⃣ Parameter Update

After gradients are calculated, an optimization algorithm updates the network parameters.

The general idea:

```text
Old Weight

      ↓

Apply Gradient Update

      ↓

New Weight
```

The update depends on:

- Gradient
- Learning Rate
- Optimizer

Common optimizers:

- SGD (Stochastic Gradient Descent)
- Adam
- RMSprop

---


# 🔢 Epoch

An **Epoch** means one complete pass through the entire training dataset.

Example:

Dataset: `10,000 samples`

Training:

```text
Epoch 1:  The model sees all 10,000 samples once


Epoch 2:  The model sees all 10,000 samples again


Epoch 3:  The model sees all 10,000 samples again
```

More epochs allow the model to learn more patterns, but too many epochs may cause overfitting.

---

# 🔄 Relationship Between Batch, Iteration, and Epoch

These concepts are closely related:

## Batch

A small portion of the dataset.

---

## Iteration (Step)

One update of the model parameters using one batch.

Example:

```text
One batch

↓

Forward Propagation

↓

Backpropagation

↓

Weight Update

=

One Iteration
```

---

## Epoch

One complete pass through the whole dataset.

Example:

Dataset:  `10,000 samples`

Batch size:  `100 samples`

Number of iterations:  `10000 \div 100 = 100`

Therefore:  `1 Epoch = 100 Iterations`
